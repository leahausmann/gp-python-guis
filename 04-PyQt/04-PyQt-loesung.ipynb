```python
from PyQt6.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QLineEdit, QHBoxLayout
import sys


class PyQtExerciseWindow(QWidget):
    def __init__(self):
        super().__init__()

        self.setWindowTitle("PyQt Übung")
        self.label = QLabel("Gib einen Text ein:")
        self.text_input = QLineEdit()
        self.button = QPushButton("Anzeigen")

        layout = QHBoxLayout()
        layout.addWidget(self.label)
        layout.addWidget(self.text_input)
        layout.addWidget(self.button)
        self.setLayout(layout)

        self.button.clicked.connect(self.update_label)

    def update_label(self):
        text = self.text_input.text().strip()
        self.label.setText(text if text else "Bitte gib etwas ein.")


def main():
    app = QApplication(sys.argv)
    window = PyQtExerciseWindow()
    window.show()
    sys.exit(app.exec())


if __name__ == "__main__":
    main()
```
