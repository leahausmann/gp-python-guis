## 1. Grundprinzip eines PyQt-Programms

Jede PyQt-Anwendung besteht im Kern aus drei Bausteinen:

1. `QApplication` – repräsentiert die Anwendung selbst  
   → kümmert sich darum, dass die GUI läuft  
2. `Widgets` – das sind die sichtbaren Elemente  
   → z. B. Fenster, Buttons, Labels, Eingabefelder  
3. `Event-Loop` – eine Schleife, die auf Benutzeraktionen wartet  
   → z. B. Klicks, Tastatureingaben, Fenster schließen

Ein minimales Beispiel sieht so aus:



```python
from PyQt6.QtWidgets import QApplication, QWidget

app = QApplication([])           # erstellt die Anwendung
window = QWidget()               # erstellt ein leeres Fenster
window.setWindowTitle("PyQt Demo")
window.show()                    # zeigt das Fenster an
app.exec()                       # startet die Event-Loop
```

## 2. Widgets und Layouts
In PyQt ist jedes sichtbare Element ein Widget:

- `QLabel` – Text anzeigen
- `QPushButton` – Button
- `QLineEdit` – einzeiliges Eingabefeld
- `QVBoxLayout` / QHBoxLayout – Layouts zur Anordnung von Widgets



Statt feste Pixelpositionen zu verwenden, arbeiten wir mit Layouts.
Das Layout entscheidet, wo Widgets im Fenster erscheinen und sorgt dafür, dass sich alles automatisch mit der Fenstergröße anpasst.

→ Beispiel mit Label, Button und vertikalem Layout:

```python
from PyQt6.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QVBoxLayout

app = QApplication([])

window = QWidget()
window.setWindowTitle("Widgets und Layouts")

label = QLabel("Hallo, ich bin ein Label")
button = QPushButton("Klick mich")

layout = QVBoxLayout()      # vertikales Layout
layout.addWidget(label)     # Label oben
layout.addWidget(button)    # Button darunter

window.setLayout(layout)    # Layout dem Fenster zuweisen
window.show()

app.exec()
```

## 3. Signale und Slots – Reaktionen auf Benutzeraktionen

- Das wichtigste Funktionsprinzip in PyQt sind Signale und Slots:
- Ein Signal wird ausgelöst, wenn etwas passiert
    → z. B. ein Button wird angeklickt (button.clicked)
- Ein Slot ist eine Funktion, die auf dieses Signal reagiert
- Wir verbinden beides mit `connect()`:

```python
button.clicked.connect(meine_funktion)
```

Wenn der Button geklickt wird, ruft PyQt automatisch **meine_funktion()** auf.

**Beispiel: Button ändert Label-Text**

```python
from PyQt6.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QVBoxLayout

def on_button_clicked():
    label.setText("Der Button wurde geklickt!")

app = QApplication([])

window = QWidget()
window.setWindowTitle("Signale und Slots")

label = QLabel("Bitte klicke auf den Button")
button = QPushButton("Klick mich")

button.clicked.connect(on_button_clicked)  # Signal mit Funktion verbinden

layout = QVBoxLayout()
layout.addWidget(label)
layout.addWidget(button)
window.setLayout(layout)

window.show()
app.exec()
```

## 4. Datenfluss zwischen Widgets

Oft sollen Eingaben von einem Widget in einem anderen angezeigt werden.
Das funktioniert, indem eine Slot-Funktion den Benutzertext ausliest und weiterverarbeitet.


Beispiel: Text aus Eingabefeld übernehmen

```python
from PyQt6.QtWidgets import QApplication, QWidget, QLabel, QPushButton, QLineEdit, QVBoxLayout

def update_label():
    eingabe = text_input.text().strip()
    if eingabe:
        label.setText(f"Du hast eingegeben: {eingabe}")
    else:
        label.setText("Bitte gib etwas ein.")

app = QApplication([])
window = QWidget()
window.setWindowTitle("Datenfluss zwischen Widgets")

label = QLabel("Gib einen Text ein und drücke den Button:")
text_input = QLineEdit()
button = QPushButton("Übernehmen")

button.clicked.connect(update_label)

layout = QVBoxLayout()
layout.addWidget(label)
layout.addWidget(text_input)
layout.addWidget(button)
window.setLayout(layout)

window.show()
app.exec()
```

| Schritt | Aktion                                  |
| ------- | --------------------------------------- |
| 1       | Benutzer tippt Text ins Textfeld        |
| 2       | Klick erzeugt Signal                    |
| 3       | Slot liest den Text und verarbeitet ihn |
| 4       | Label zeigt Ergebnis an                 |


Damit entsteht eine interaktive Logik:
Der Nutzer bestimmt, **was** passiert — PyQt führt es aus.
