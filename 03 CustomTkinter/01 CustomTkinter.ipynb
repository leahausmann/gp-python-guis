# Einführung in CustomTkinter 

„Wenn man mit Python eine moderne Oberfläche programmieren möchte, stößt man schnell an die Grenzen von Tkinter.  
Deshalb gibt es **CustomTkinter**: eine moderne, stilvolle Erweiterung mit Dark-Mode, Themes und einfachen Widgets, die viel besser aussehen und leichter zu benutzen sind.“

---

## 1. Installation

CustomTkinter mit **pip install customtkinter** installieren
Anschließend CustomTkinter mit **import customtkinter as ctk** in die Programmierumgebung importieren


## 2. Hauptfenster erstellen

app = ctk.CTk()
app.title("CustomTkinter Demo")
app.geometry("400x200")


## 3. Das wichtigste: Mainloop erstellen

app.mainloop()


## 4. Das Hauptfenster bearbeiten:

- Erscheinungsmodus einstellen --> ctk.set_appearance_mode("Dark"); ctk.set_appearance_mode("light"); ctk.set_appearance_mode("system")
- Widgets hinzufügen: Button --> button = ctk.CTkButton(app, text = "TEXT")
- Position des Widgets mit "widget_name.pack(pady=x)" einstellen --> button.pack(pady=10)


```python
import customtkinter as ctk



app = ctk.CTk()  
app.title("CustomTkinter Demo")
app.geometry("400x200")


ctk.set_appearance_mode("dark")


button = ctk.CTkButton(app, text="Klick mich") 
button.pack (pady=10) 


name_entry = ctk.CTkEntry(app, placeholder_text="Dein Name")
name_entry.pack(pady=30)

checkbox = ctk.CTkCheckBox(app, text="Ich stimme zu")
checkbox.pack(pady=10)


app.mainloop()
```
