# Kivy

Kivy ist eine Python-Bibliothek, mit der man GUIs und Apps auf mehreren Plattformen entwickeln kann.
Es ist besonders darauf ausgelegt, Touch-Gesten zu erkennen, wie z.B.:

Wischen, Zoomen mit zwei Fingern, Drücken oder Antippen

Funktioniert plattformübergreifend auf Windows, Linux, macOS, Android und iOS.

## 1. Installation

Kivy über pip installieren: **pip install kivy**

## 2. Beispiel




```python
from kivy.app import App
from kivy.uix.label import Label
from kivy.uix.button import Button
from kivy.uix.textinput import TextInput
from kivy.uix.boxlayout import BoxLayout


class MyApp(App):
    def build(self):
        # Hauptlayout (alles untereinander)
        layout = BoxLayout(orientation="vertical", padding=20, spacing=15)

        # Widgets erstellen
        title = Label(
            text="Willkommen in meiner Kivy App!",
            font_size=30
        )

        input_field = TextInput(
            hint_text="Gib etwas ein...",
            font_size=20,
            size_hint=(1, 0.4)
        )

        button = Button(
            text="Klick mich!",
            font_size=24
        )

        # Widgets ins Layout einfügen
        layout.add_widget(title)
        layout.add_widget(input_field)
        layout.add_widget(button)

        return layout



MyApp().run()
```

```python
# Lösungsvorschlag

from kivy.app import App
from kivy.uix.label import Label
from kivy.uix.button import Button
from kivy.uix.textinput import TextInput
from kivy.uix.boxlayout import BoxLayout


class MyApp(App):
    def build(self):
        # Hauptlayout (alles untereinander)
        layout = BoxLayout(orientation="vertical", padding=30, spacing=30)

        # Widgets erstellen
        title = Label(
            text="KIVY ÜBUNG",
            font_size=30
        )

        input_field = TextInput(
            hint_text="Gib etwas ein...",
            font_size=30,
            size_hint=(1, 0.4)
        )

        button = Button(
            text="GESCHAFFT",
            font_size=30
        )

        # Widgets ins Layout einfügen
        layout.add_widget(title)
        layout.add_widget(input_field)
        layout.add_widget(button)

        return layout



MyApp().run()
```
