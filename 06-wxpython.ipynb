### Allgemein
WxPython ist ein GUI-Framework für Python, das auf dem `wxWidgets`-Toolkit basiert. Es ist ähnlich zu den anderen Frameworks, allerdings mit eigenem API-Stil.

**<u>Ziel:</u>** `Native Look & Feel` auf verschiedenen Betriebssystemen (Windows, macOS, Linux).

**<u>Grundidee:</u>**
- `wx.App`-Instanz als Anwendung
- ein Hauptfenster erbt meist von `wx.Frame`
- im Hauptfenster können andere GUI-Elemente platziert werden (`Buttons`, `Textfelder`, `Layouts` etc.)

---

### Aufbau eines Programms mit wxPython

Der typische Aufbau sieht so aus:

```python
import wx

class MyFrame(wx.Frame):
    def __init__(self):
        super().__init__(parent=None, title="Mein Fenster")
        # Hier Widgets/Layouts einrichten

if __name__ == "__main__":
    app = wx.App()
    frame = MyFrame()
    frame.Show()
    app.MainLoop()
```


---

### Beispiel

````python
import wx

class MyFrame(wx.Frame):
    def __init__(self):
        super().__init__(parent=None, title="Mein erstes wxPython-Fenster")
        
        panel = wx.Panel(self)
        
        text = wx.StaticText(panel, label="Willkommen bei wxPython!", pos=(20, 20))
        
        self.button = wx.Button(panel, label="Klick mich", pos=(20, 60))
        self.button.Bind(wx.EVT_BUTTON, self.on_button_click)
        
        self.SetSize((400, 200))
        self.Centre()
    
    def on_button_click(self, event):
        wx.MessageBox("Button wurde geklickt!", "Info", wx.OK | wx.ICON_INFORMATION)

if __name__ == "__main__":
    app = wx.App()
    frame = MyFrame()
    frame.Show()
    app.MainLoop()
````

### Erklärung

- `wx.App()` erstellt die Anwendung.
- `MyFrame` erbt von `wx.Frame` und stellt das Hauptfenster dar.
- `wx.Panel` ist eine Art Container innerhalb des Fensters.
- `wx.StaticText` zeigt einen Text an.
- `wx.Button` erzeugt einen Button.
- Mit `Bind(wx.EVT_BUTTON, self.on_button_click)` verknüpfen wir den Button mit einer Funktion.
- `wx.MessageBox(...)` zeigt ein kleines Meldungsfenster an.


---

### Layout

Statt mit `pos=(x, y)` absolute Positionen zu setzen, nutzt man in wxPython oft **Sizers** (Layout-Manager).  

#### Beispiel:
`wx.BoxSizer` für vertikale/horizontale Anordnung.


````python
import wx

class SizerFrame(wx.Frame):
    def __init__(self):
        super().__init__(parent=None, title="wxPython mit Sizers")
        
        panel = wx.Panel(self)
        
        vbox = wx.BoxSizer(wx.VERTICAL)
        
        self.text = wx.StaticText(panel, label="Noch nicht geklickt")
        vbox.Add(self.text, flag=wx.ALL | wx.ALIGN_CENTER_HORIZONTAL, border=10)
        
        button = wx.Button(panel, label="Ändere Text")
        vbox.Add(button, flag=wx.ALL | wx.ALIGN_CENTER_HORIZONTAL, border=10)
        
        button.Bind(wx.EVT_BUTTON, self.on_button_click)
        
        panel.SetSizer(vbox)
        
        self.SetSize((400, 200))
        self.Centre()
    
    def on_button_click(self, event):
        self.text.SetLabel("Button wurde geklickt!")

if __name__ == "__main__":
    app = wx.App()
    frame = SizerFrame()
    frame.Show()
    app.MainLoop()
````
