## Abschluss & Vergleich der GUI-Bibliotheken

- Heute vergleichen wir die vorgestellten GUI-Bibliotheken.
- Ziel: Überblick geben, damit klar wird, welche Bibliothek wofür geeignet ist.

<br>

---

##  Tkinter: der einfache Einstieg

- Einsteigerfreundliche Bibliothek
- Direkt in Python integriert → keine Installation nötig
- Ideal für kleine Programme und schnelle Ergebnisse
- Perfekt für alle, die gerade anfangen und einfache GUIs bauen wollen

<br>

---

##  CustomTkinter: modern und hübsch

- Verbesserte, moderne Variante von Tkinter
- Runde Buttons, Dark Mode, ansprechendes Design
- Code fast identisch mit Tkinter
- Sehr geeignet für Schul- und Uni-Projekte
- Gute Wahl, wenn die Oberfläche moderner wirken soll

<br>

---

##  PyQt: professionelle Anwendungen

- Sehr umfangreiche und professionelle Bibliotheken
- Für komplexe Desktop-Anwendungen geeignet
- Viele echte Programme basieren auf Qt
- Integrierter GUI-Designer → Drag-and-Drop-Erstellung
- Beste Wahl für große, langfristige oder industrielle Projekte

<br>

---

##  Kivy: perfekt für mobile Geräte und Touch

- Speziell für mobile Apps und Touch-Oberflächen
- Unterstützt Gesten: Wischen, Zoomen, Drücken
- Läuft auf vielen Plattformen: Windows, Android, iOS
- Einmal programmieren → auf mehreren Geräten nutzbar
- Ideal für Apps, Spiele und interaktive Anwendungen

<br>

---

##  wxPython: native Systemoberflächen

- Nutzt echte System-Widgets des Betriebssystems
- Programme sehen unter Windows wie Windows-Apps aus
- Unter macOS wie macOS-Apps, unter Linux wie Linux-Apps
- Fügt sich optisch perfekt ins System ein
- Sehr geeignet für systemnahe, professionelle Desktop-Programme

<br>

---

##  Fazit:

- **Tkinter** → einfacher Einstieg  
- **CustomTkinter** → modernere und schönere Oberflächen  
- **PyQt / PySide** → professionelle & komplexe Software  
- **Kivy** → mobile & touchbasierte Anwendungen  
- **wxPython** → native und systemnahe GUIs  

- Keine Bibliothek ist „die beste“ – jede hat ihr Einsatzgebiet.
- Python bietet für jedes GUI-Projekt die passende Lösung.
- Wichtig: Vorher überlegen, was man entwickeln möchte → passende Bibliothek wählen.
