# GUI – Graphical User Interface

---

## 1. Was ist eine GUI?

<br>

- **GUI** = Graphical User Interface

- eine **grafische Benutzerschnittstelle**, die Programme über sichtbare Bedienelemente wie **Fenster, Icons, Buttons und Menüs** steuerbar macht

- statt Eingabe von Textbefehlen → Interaktion mit einem System über **Maus, Touchpad oder Touchscreen**

<br>

---

## 2. Grundlagen und Bestandteile

<br>

- basiert auf einem Fenstersystem im Grafikmodus, das die visuellen Elemente darstellt

<br>

### WIMP-Modell

<br>

- **Windows:** Bereiche, in denen Programme laufen

- **Icons:** kleine grafische Darstellungen von Funktionen oder Dateien

- **Menus:** zeigen auswählbare Befehle oder Einstellungen

- **Pointer:** Zeiger, der die Eingaben steuert  

<br>

---

## 3. Vor- und Nachteile

<br>

### Vorteile

- äußerst **benutzerfreundlich** und **intuitiv**

- keine besonderen Kenntnisse für die Bedienung notwendig

<br>

### Nachteile

- höherer **Ressourcenverbrauch** → mögliche Beeinträchtigung der Leistung

- **aufwendigere** und **zeitintensivere** Entwicklung

<br>

---

## 4. Historie bis heute

<br>

- Einführung in den **frühen 1970er-Jahren**

- ursprünglich mussten Nutzer Befehle über die **Kommandozeile** eingeben

- heute prägen GUIs nahezu alle Geräte mit visueller Benutzeroberfläche (z. B. Computer, Smartphones, Tablets, Kiosksysteme)
