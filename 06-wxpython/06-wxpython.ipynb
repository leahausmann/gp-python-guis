### Was ist wxPython?
- ein Python-Wrapper für das C++-Framework `wxWidgets`
- `wxWidgets` erzeugt native Fenster und Widgets
- es werden die nativen GUI-Elemente des Betriebssystems verwendet:

    | Betriebssystem | Verwendete native APIs   |
    | -------------- | ------------------------- |
    | Windows        | Win32 / UWP Controls      |
    | macOS          | Cocoa (Objective-C/Swift) |
    | Linux          | GTK3/GTK4                 |

- `wxPython` verbindet Python-Code mit den C++-Elementen von `wxWidgets`

---

### Wie ist wxPython aufgebaut?
1. `Python`-Code: hier definieren wir unsere Fenster und Widgets
2. `wxPython`-Wrapper: Übersetzt den Python-Code in C++
3. `wxWidgets` (C++): native Kontrolle über das Betriebssystem, Events, Rendering
4. `Betriebssystem`: erzeugt Fenster und Widgets


```python
frame = wx.Frame(None, title="Fenster")
```

#### Was passiert hier?
1. wxPython erstellt ein Python-Objekt wx.Frame
2. Die wxPython-Schicht ruft den passenden Constructor in wxWidgets (C++) auf
3. wxWidgets ruft die native OS-API auf (Win32/Cocoa/GTK)
4. Das Betriebssystem erzeugt ein echtes Fenster
5. wxWidgets merkt sich das native Handle (z. B. HWND unter Windows)
6. wxPython speichert einen Zeiger auf das C++-Objekt
7. Dein Python-Objekt steuert jetzt ein echtes C++-System-Fenster

---

### Layout / Sizers
- `wxPython` rendert UI nicht absolut (Pixel für Pixel), sondern mit Layouts, sogenannten Sizers: `BoxSizer`, `GridSizer`, `FlexGridSizer`, `GridBagSizer` und mehr
- Fenster skalieren automatisch, Widgets passen sich Heigh-DPI und Retina geräteunabhängig an, UI bleibt flexibel und responsiv

---

### Warum wird wxPython verwendet?
- Native UI
- Sehr stabil (wxWidgets existiert seit Jahrzehnten)
- Plattformunabhängig
- schnell genug, da im C++-Teil gerendert wird
