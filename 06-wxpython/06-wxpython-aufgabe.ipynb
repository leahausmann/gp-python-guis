## Beispielcode

```python
import wx

class MyFrame(wx.Frame):
    def __init__(self):
        super().__init__(None, title="wxPython Beispiel")

        panel = wx.Panel(self)
        button = wx.Button(panel, label="Klick mich")

        button.Bind(wx.EVT_BUTTON, self.on_click)

        self.Show()

    def on_click(self, event):
        wx.MessageBox("Button wurde geklickt!")

app = wx.App()
MyFrame()
app.MainLoop()
```
