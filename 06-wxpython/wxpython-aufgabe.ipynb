## Wichtige Befehle/Funktionen:

### Widgets

```python
# Textfeld
text = wx.TextCtrl(panel)

# Label
label = wx.StaticText(panel, label="...")

# Button
button = wx.Button(panel, label="OK")
```

### Methoden

```python
# Text auslesen
text.GetValue()

# Label setzen
label.SetLabel("Text")
```

### Events

```python
# Button-Klick binden
button.Bind(wx.EVT_BUTTON, self.on_click)

# Event-Handler
def on_click(self, event):
    ...
```

### Layout

```python
# Vertikaler BoxSizer
sizer = wx.BoxSizer(wx.VERTICAL)
panel.SetSizer(sizer)

# Elemente hinzufügen
sizer.Add(widget, 0, wx.ALL | wx.EXPAND, 5)
```
