```python
import wx

class MyFrame(wx.Frame):
    def __init__(self):
        # Fenster erstellen
        super().__init__(None, title="")

        # Panel als Arbeitsfläche
        panel = wx.Panel(self)

        # TODO: Widgets einfügen
        self.text_ctrl = wx.TextCtrl(panel)
        self.label = wx.StaticText(panel, label="Hier erscheint der Text")
        self.button = wx.Button(panel, label="Anzeigen")

        # TODO: Den Button mit einer Event-Methode verbinden
        self.button.Bind(wx.EVT_BUTTON, self.on_button_click)
        
        # TODO: Hier den Sizer einfügen
        sizer = wx.BoxSizer(wx.VERTICAL)
        sizer.Add(self.text_ctrl, 0, wx.EXPAND | wx.ALL, 5)
        sizer.Add(self.button, 0, wx.ALL, 5)
        sizer.Add(self.label, 0, wx.ALL, 5)
        panel.SetSizer(sizer)

        self.Centre()  # Fenster zentrieren
        self.Show()

    # TODO: Hier die Methode einfügen, die bei Klick auf den Button aufgerufen wird
    def on_button_click(self, event):
        text = self.text_ctrl.GetValue()
        self.label.SetLabel(text)
   
if __name__ == "__main__":
    app = wx.App()
    frame = MyFrame()
    app.MainLoop()
```
