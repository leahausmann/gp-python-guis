


# Tkinter  
## Die Standard-GUI-Bibliothek von Python  

---

# Was ist Tkinter?

Tkinter ist die **integrierte GUI-Bibliothek** von Python.  
Damit kann man Programme erstellen, die echte Fenster und Steuerelemente besitzen.

Vorteile von Tkinter:

- Keine Installation nötig – direkt in Python enthalten  
- Ideal für Einsteiger  
- Plattformübergreifend (Windows, macOS, Linux)  
- Viele fertige Widgets für Buttons, Labels, Eingabefelder, Menüs usw.

---

# Wie funktioniert Tkinter technisch?

Tkinter ist ein **Wrapper für die Tcl/Tk-Bibliothek**:


- Tkinter übersetzt Python-Befehle in Tcl-Kommandos  
- Tcl/Tk erstellt das Fenster und die Widgets  
- Das Betriebssystem zeigt das Fenster auf dem Bildschirm an  

So steuert dein Python-Code ein echtes Fenster, ohne dass du direkt das Betriebssystem ansprechen musst.

---

# Aufbau eines Tkinter-Programms

Ein typisches Tkinter-Programm besteht aus:

1. Hauptfenster erstellen  
2. Layout festlegen (Pack, Grid oder Place)  
3. Widgets hinzufügen (Button, Label, Entry etc.)  
4. mainloop() starten, damit das Fenster aktiv bleibt

---


# Beispiel: Tkinter mit Klasse

```python

import tkinter as tk
from tkinter import messagebox

class MyWindow:
    def __init__(self):
        self.root = tk.Tk()
        self.root.title("Tkinter Beispiel")

        # Frame als Container
        frame = tk.Frame(self.root)
        frame.pack(padx=20, pady=20)

        # Button
        button = tk.Button(frame, text="Klick mich", command=self.on_click)
        button.pack()

        self.root.mainloop()

    def on_click(self):
        messagebox.showinfo("Info", "Button wurde geklickt!")

MyWindow()
```
